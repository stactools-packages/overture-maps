# stactools-overture-maps

This is a small notebook demonstrating use of this package and its generated items and collections.
It is intended to be executed from the root of this repository, since it relies on files that have been checked into this repo.

## Example items and collections

This package provides some example data, in the `examples/` directory.
You can load them with [pystac](https://github.com/stac-utils/pystac).
The item includes a custom attribute that is specific to this package.

```python
from pathlib import Path

from IPython.display import display
from pystac import Catalog

import stactools.overture_maps.stac
from stactools.overture_maps.metadata import StorageBackend, Theme

root = Path().cwd().parent
catalog = Catalog.from_file(str(root / "examples" / "catalog.json"))
display(catalog)
collection = next(catalog.get_children())
display(collection)
item = next(collection.get_items())
display(item)
print(f"This is the release: {item.properties['version']}")
```

## Creating items and collections

Items can be created from an href to a single tif file.
Items are added to collections using the standard pystac interface.

```python
href = "https://overturemapswestus2.blob.core.windows.net/release/2024-12-18.0/theme=addresses/type=address/part-00000-de803747-d78d-4060-b3da-da6dcd5dbab8-c000.zstd.parquet"
item = stactools.overture_maps.stac.create_item(href)
collection = stactools.overture_maps.stac.create_collection(
    theme=Theme.ADDRESSES,
    storage_backend=StorageBackend.AZURE,
    latest_release="2024-12-18.0",
)
collection.add_item(item)
display(collection)
```
